```python
import pandas as pd

url = "https://storage.googleapis.com/download.tensorflow.org/data/creditcard.csv"
df = pd.read_csv(url)

print(df.shape)
print(df.columns.tolist())
print(df["Class"].value_counts())
print("Fraud rate:", round(df["Class"].mean() * 100, 3), "%")
print("Missing values:", df.isnull().sum().sum())
print(df.groupby("Class")["Amount"].describe())
```

```python
# 1. Fraud rate by hour of day
df["Hour"] = (df["Time"] // 3600) % 24
print(df.groupby("Hour")["Class"].mean().mul(100).round(3))

# 2. Which hidden features separate fraud from genuine the most
v_cols = [f"V{i}" for i in range(1, 29)]
fraud_mean = df[df["Class"] == 1][v_cols].mean()
genuine_mean = df[df["Class"] == 0][v_cols].mean()
gap = ((fraud_mean - genuine_mean) / df[v_cols].std()).abs()
print(gap.sort_values(ascending=False).head(8).round(2))

# 3. Fraud rate for tiny transactions (2 or less) vs others
df["small_txn"] = (df["Amount"] <= 2).astype(int)
print(df.groupby("small_txn")["Class"].mean().mul(100).round(3))
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, precision_score, recall_score

# 1. Build features from what we found
df["log_amount"] = np.log1p(df["Amount"])
df["night"] = df["Hour"].between(1, 7).astype(int)
features = [f"V{i}" for i in range(1, 29)] + ["log_amount", "night", "small_txn"]

X = df[features]
y = df["Class"]

# 2. Split 70/30, keeping fraud rate equal in both
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)

scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)
print("Test frauds:", y_test.sum(), "out of", len(y_test))

# 3. Train logistic regression that pays extra attention to fraud
lr = LogisticRegression(max_iter=1000, class_weight="balanced")
lr.fit(X_train_s, y_train)

# 4. Flag as fraud if probability is 0.5 or more
lr_prob = lr.predict_proba(X_test_s)[:, 1]
lr_pred = (lr_prob >= 0.5).astype(int)

print(confusion_matrix(y_test, lr_pred))
print("Precision:", round(precision_score(y_test, lr_pred), 3))
print("Recall:", round(recall_score(y_test, lr_pred), 3))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, average_precision_score

# 1. Train XGBoost (trees don't need scaled data)
xgb = XGBClassifier(n_estimators=300, max_depth=4, learning_rate=0.1,
                    eval_metric="aucpr", random_state=42)
xgb.fit(X_train, y_train)
xgb_prob = xgb.predict_proba(X_test)[:, 1]

# 2. Compare both models side by side
for name, prob in [("Logistic", lr_prob), ("XGBoost", xgb_prob)]:
    pred = (prob >= 0.5).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, pred).ravel()
    print(f"{name}: ROC-AUC {roc_auc_score(y_test, prob):.3f} | "
          f"PR-AUC {average_precision_score(y_test, prob):.3f} | "
          f"caught {tp}/{tp + fn} | good blocked {fp}")

# 3. Which features XGBoost relied on most
imp = pd.Series(xgb.feature_importances_, index=features)
print(imp.sort_values(ascending=False).head(8).round(3))
```

```python
from sklearn.metrics import precision_recall_curve
import matplotlib.pyplot as plt

FRAUD_COST = 150   # assumed loss per missed fraud (EUR)
BLOCK_COST = 10    # assumed cost per good customer wrongly blocked (EUR)

def evaluate(prob, t):
    pred = (prob >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, pred).ravel()
    return {"threshold": round(float(t), 2), "caught": int(tp), "missed": int(fn),
            "good_blocked": int(fp),
            "precision": round(tp / (tp + fp), 3) if tp + fp else 0,
            "recall": round(tp / (tp + fn), 3),
            "cost_eur": int(fn * FRAUD_COST + fp * BLOCK_COST)}

# 1. XGBoost at different thresholds
table = pd.DataFrame([evaluate(xgb_prob, t) for t in
                      [0.01, 0.02, 0.05, 0.1, 0.2, 0.3, 0.5, 0.7, 0.9]])
print(table.to_string(index=False))

# 2. Best threshold for each model, vs no model
grid = np.arange(0.01, 1.0, 0.01)
for name, prob in [("Logistic", lr_prob), ("XGBoost", xgb_prob)]:
    best = min((evaluate(prob, t) for t in grid), key=lambda r: r["cost_eur"])
    print(name, "best:", best)
print("No model (all fraud passes):", int(y_test.sum() * FRAUD_COST))

# 3. Precision-recall chart
for name, prob in [("Logistic", lr_prob), ("XGBoost", xgb_prob)]:
    p, r, _ = precision_recall_curve(y_test, prob)
    plt.plot(r, p, label=f"{name} (PR-AUC {average_precision_score(y_test, prob):.2f})")
plt.xlabel("Recall (fraud caught)")
plt.ylabel("Precision (flags that are real fraud)")
plt.title("Precision-Recall curve")
plt.legend()
plt.show()
```
